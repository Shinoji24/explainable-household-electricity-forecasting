```python
import pandas as pd
import numpy
```

```python
df = pd.read_csv(
    "household_power_consumption.txt",
    sep=";",
    na_values=["?"],
    # nrows = 5000,
    low_memory=False
)
```

```python
# Create a separate testing dataset
test_df = df.copy()

# Introduce 5 missing values in the first 5 rows
test_df.iloc[0:5, test_df.columns.get_loc("Global_active_power")] = pd.NA

# Save as a separate CSV file
test_df.to_csv("test_missing_values.csv", index=False)
```

```python
# Create a copy of the original dataset
test_invalid_df = df.copy()

# Remove the required target column
test_invalid_df = test_invalid_df.drop(
    columns=["Global_active_power"]
)

# Save it as a separate CSV file
test_invalid_df.to_csv(
    "test_missing_column.csv",
    index=False
)
```

```python
# Select 10 rows from the original dataset
test_duplicate_df = df.head(10).copy()

# Duplicate the first row
test_duplicate_df = pd.concat(
    [test_duplicate_df, test_duplicate_df.iloc[[0]]],
    ignore_index=True
)

# Save as a separate CSV
test_duplicate_df.to_csv(
    "test_duplicate_timestamps.csv",
    index=False
)
```

```python
# Select enough rows to cover multiple hours
test_duplicate_df = df.head(3000).copy()

# Duplicate the first row
test_duplicate_df = pd.concat(
    [test_duplicate_df, test_duplicate_df.iloc[[0]]],
    ignore_index=True
)

# Save separately
test_duplicate_df.to_csv(
    "test_duplicate_timestamps.csv",
    index=False
)
```
